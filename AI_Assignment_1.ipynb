## Data Manipulation with Pandas

In this section, I will create a small dataset using a Python dictionary.

The dataset contains information about five students:

- **Name:** Student name
- **Study_Hours:** Number of hours studied per week
- **Quiz_Score:** Score received on a quiz

I will convert this dictionary into a Pandas DataFrame and display it.

```python
# Import the libraries needed for this assignment

# pandas is used to create and work with tables of data (DataFrames)
import pandas as pd

# matplotlib is used to create charts and graphs
import matplotlib.pyplot as plt

# seaborn makes statistical charts easier and more attractive
import seaborn as sns

# This command makes charts appear directly below the code cells
%matplotlib inline

# Check that pandas was imported successfully
print("Pandas imported successfully!")
```

```python
# Create a sample dataset using a dictionary
student_data = {
    "Name": ["Ali", "Sara", "John", "Maria", "David"],
    "Study_Hours": [5, 8, 3, 10, 6],
    "Quiz_Score": [70, 85, 60, 95, 78]
}

# Convert the dictionary into a Pandas DataFrame
student_df = pd.DataFrame(student_data)

# Display the DataFrame
student_df
```

### Description of the Dataset

The dataset contains information about five students.

It has three columns:

1. **Name** — The name of each student.
2. **Study_Hours** — The number of hours each student studied during the week.
3. **Quiz_Score** — The score each student received on the quiz.

This dataset can help us understand whether students who study more hours tend to receive higher quiz scores.

```python
# Display the first 5 rows of the DataFrame
print("First 5 rows:")
display(student_df.head())

# Display basic information about the DataFrame
print("\nDataset information:")
student_df.info()

# Display summary statistics for numerical columns
print("\nSummary statistics:")
display(student_df.describe())
```

## Data Visualization with Matplotlib and Seaborn

In this section, I will create a bar chart to compare the quiz scores of the students.

A bar chart is useful when we want to compare values between different categories, such as student names.

```python
# Set the size of the chart
plt.figure(figsize=(8, 5))

# Create a bar chart
# x-axis: student names
# y-axis: quiz scores
sns.barplot(
    x="Name",
    y="Quiz_Score",
    data=student_df,
    palette="viridis"
)

# Add a title and axis labels
plt.title("Quiz Scores of Students")
plt.xlabel("Student Name")
plt.ylabel("Quiz Score")

# Display the chart
plt.show()
```

### Explanation of the Visualization

The bar chart shows the quiz scores of five students.

From the chart, we can see that:

- **Maria** has the highest quiz score.
- **John** has the lowest quiz score.
- **Sara** also has a high score.

The chart suggests that students with more study hours may have higher quiz scores. For example, Maria studied 10 hours and received the highest score, while John studied only 3 hours and received the lowest score.

However, this is a very small sample dataset, so we cannot make a strong conclusion from it.

### Using Magic Commands

The `%timeit` command measures how long a piece of Python code takes to run.

In this example, I will measure the execution time of a list comprehension that creates a list of squared numbers.

```python
# Use %timeit to measure the execution time of a list comprehension

# This list comprehension creates a list of squares from 1 to 1000
# For example: [1, 4, 9, 16, ..., 1000000]

%timeit [x ** 2 for x in range(1000)]
```
