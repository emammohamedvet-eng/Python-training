## Initial Notebook: Multiplication Function

I will create a Jupyter notebook named `assignment2.ipynb`
inside the cloned repository.

The initial notebook will contain the required assignment heading,
a multiplication function, and a test using the values 3 and 4.

The expected result is 12.

```python
# Define the required multiplication code
multiplication_code = """def multiply_numbers(a, b):
    # Return the product of two numbers
    return a * b

print(multiply_numbers(3, 4))
"""

# Run a code cell and capture its printed output
def make_executed_cell(source, execution_count):
    output_buffer = io.StringIO()

    with contextlib.redirect_stdout(output_buffer):
        exec(source, {})

    output_text = output_buffer.getvalue()

    cell = nbformat.v4.new_code_cell(source)
    cell.execution_count = execution_count

    if output_text:
        cell.outputs = [
            nbformat.v4.new_output(
                output_type="stream",
                name="stdout",
                text=output_text
            )
        ]

    return cell

# Define the file location
notebook_path = repo_path / "assignment2.ipynb"

# Avoid overwriting work if this cell is accidentally repeated
if notebook_path.exists():
    raise FileExistsError(
        "assignment2.ipynb already exists. "
        "Do not overwrite it without checking your progress."
    )

# Create the notebook
assignment_nb = nbformat.v4.new_notebook()

assignment_nb.metadata["kernelspec"] = {
    "display_name": "Python 3",
    "language": "python",
    "name": "python3"
}

# Add the required Markdown and multiplication cells
assignment_nb.cells = [
    nbformat.v4.new_markdown_cell(
        "# Git and GitHub Assignment 2\n"
        "This notebook demonstrates basic Git operations."
    ),
    make_executed_cell(multiplication_code, 1)
]

# Validate and save the notebook
nbformat.validate(assignment_nb)
nbformat.write(assignment_nb, notebook_path)

print("Created:", notebook_path.name)
print("Multiplication output:")
print(assignment_nb.cells[1].outputs[0].text)
```

```python
# Import the tools needed for authentication
import os
import subprocess
import tempfile
from pathlib import Path
from getpass import getpass

# Set your repository details
github_username = "emammohamedvet-eng"
repo_path = Path("/content/colab-git-assignment2-EM")

def authenticated_push(*push_arguments):
    # Enter the token through a hidden prompt
    token = getpass("Enter your GitHub PAT (hidden): ").strip()

    if not token:
        print("No token entered. Run the push cell again.")
        return

    try:
        with tempfile.TemporaryDirectory() as temp_dir:
            # Create a temporary helper outside the repository
            helper_path = Path(temp_dir) / "git_askpass.sh"

            helper_path.write_text(
                '#!/bin/sh\n'
                'case "$1" in\n'
                '  *Username*) printf "%s\\n" "$ASSIGNMENT_GH_USER" ;;\n'
                '  *Password*) printf "%s\\n" "$ASSIGNMENT_GH_TOKEN" ;;\n'
                'esac\n'
            )

            helper_path.chmod(0o700)

            # Supply credentials only to the Git subprocess
            environment = os.environ.copy()
            environment.update({
                "GIT_ASKPASS": str(helper_path),
                "GIT_TERMINAL_PROMPT": "0",
                "ASSIGNMENT_GH_USER": github_username,
                "ASSIGNMENT_GH_TOKEN": token
            })

            # Capture Git's output so we can see the actual problem
            result = subprocess.run(
                [
                    "git", "-c", "credential.helper=",
                    "push", *push_arguments
                ],
                cwd=repo_path,
                env=environment,
                capture_output=True,
                text=True
            )

            # Hide the token if it unexpectedly appears in output
            output = (
                (result.stdout or "") + "\n" + (result.stderr or "")
            ).replace(token, "[TOKEN HIDDEN]")

            print(output.strip())

            if result.returncode == 0:
                print("\nPush completed successfully.")
            else:
                print(f"\nPush failed: exit code {result.returncode}.")

    finally:
        token = None

print("Diagnostic push helper ready.")
```

```python
# Upload the initial commit to the main branch
authenticated_push("-u", "origin", "main")
```

## Branching and Updates

I created a new branch named `feature-branch` to add a division function.

The division function checks whether the denominator is zero.
If the denominator is zero, it returns a message instead of
attempting the division.

This update will be committed and pushed separately from the
initial version on the main branch.

```python
# Locate the notebook inside the repository
notebook_path = repo_path / "assignment2.ipynb"

# Read the existing notebook
assignment_nb = nbformat.read(notebook_path, as_version=4)

# Define the required division code
division_code = """def divide_numbers(a, b):
    # Check the denominator before dividing
    if b != 0:
        return a / b
    else:
        return "Cannot divide by zero"

print(divide_numbers(10, 2))
print(divide_numbers(10, 0))
"""

# Avoid adding duplicate division cells
already_added = any(
    cell.cell_type == "code"
    and "def divide_numbers" in cell.source
    for cell in assignment_nb.cells
)

if not already_added:
    # Add an explanation
    assignment_nb.cells.append(
        nbformat.v4.new_markdown_cell(
            "## Division Function\n"
            "This function divides two numbers and handles division by zero."
        )
    )

    # Add the new Python code
    assignment_nb.cells.append(
        nbformat.v4.new_code_cell(division_code)
    )

    # Save the updated notebook
    nbformat.write(assignment_nb, notebook_path)

    print("Division function added to assignment2.ipynb.")
else:
    print("Division function is already present.")

# Test the division code
exec(division_code)
```

```python
# Review the modified files
subprocess.run(["git", "status"], check=True)

# Stage the updated notebook
subprocess.run(
    ["git", "add", "assignment2.ipynb"],
    check=True
)

# Commit the division function
subprocess.run(
    ["git", "commit", "-m", "Add division function"],
    check=True
)

# Show the latest commit
subprocess.run(
    ["git", "log", "-1", "--oneline"],
    check=True
)
```

```python
# Upload the feature branch to GitHub
authenticated_push(
    "--set-upstream",
    "origin",
    "feature-branch"
)
```

```python
# Load and validate the notebook
saved_nb = nbformat.read(notebook_path, as_version=4)
nbformat.validate(saved_nb)

# Run all Python cells in a separate namespace
test_namespace = {}

for cell in saved_nb.cells:
    if cell.cell_type == "code":
        exec(cell.source, test_namespace)

# Test the required functions
assert test_namespace["multiply_numbers"](3, 4) == 12
assert test_namespace["divide_numbers"](10, 2) == 5.0
assert (
    test_namespace["divide_numbers"](10, 0)
    == "Cannot divide by zero"
)

print("\nAll function tests passed.")

# Refresh information about branches on GitHub
subprocess.run(
    ["git", "fetch", "origin"],
    check=True
)

# Confirm both local branches match their GitHub counterparts
for branch in ["main", "feature-branch"]:
    local_commit = subprocess.check_output(
        ["git", "rev-parse", branch],
        text=True
    ).strip()

    remote_commit = subprocess.check_output(
        ["git", "rev-parse", f"origin/{branch}"],
        text=True
    ).strip()

    assert local_commit == remote_commit, (
        f"{branch} does not match GitHub. Check its push."
    )

    print(f"{branch}: local and GitHub commits match.")

# Show final status, history, and tracked files
subprocess.run(["git", "status"], check=True)

subprocess.run(
    ["git", "log", "--oneline", "--graph", "--decorate", "--all"],
    check=True
)

subprocess.run(["git", "ls-files"], check=True)
```
