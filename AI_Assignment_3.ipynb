```python
# Install required libraries (already available in Google Colab)
!pip install requests

import pickle
import sqlite3
import requests
```

```python
# 1. Data Types and Variables
book_title = "Python Programming"
book_quantity = 5
book_price = 29.99
is_available = True

print(f"Book: {book_title}, Quantity: {book_quantity}, Price: ${book_price}, Available: {is_available}")

# 2. Lists
books = ["Python Basics", "Advanced Python", "Data Science with Python"]

# TODO: Add two more books to the list
books.append("Machine Learning with Python")       # Adds one book at the end
books.insert(1, "Introduction to Python")          # Adds one book at index 1

print("Updated book list:", books)
```

```python
# 3. Loops and range()
print("\nAvailable books:")

# TODO: Use a for loop and range() to print books with their indices
for i in range(len(books)):
    print(f"{i}: {books[i]}")

# 4. User Input
new_book = input("Enter a new book title: ")

# Edge case: prevent adding an empty title
if new_book.strip():
    books.append(new_book.strip())
    print("Updated book list:", books)
else:
    print("No book title entered. The book list was not changed.")
```

```python
# 5. Functions and Conditional Statements
def check_availability(book_name):
    """Return True if a book title is in the library list; otherwise return False."""
    # Convert both values to lowercase so matching is not case-sensitive
    if book_name.strip().lower() in [book.lower() for book in books]:
        return True
    else:
        return False

# Test the function
print(f"Is 'Python Basics' available? {check_availability('Python Basics')}")
print(f"Is 'Java Programming' available? {check_availability('Java Programming')}")
```

```python
# 6. File I/O
def save_books_to_file(filename):
    """Save each book title as one line in a text file."""
    try:
        with open(filename, "w") as f:
            for book in books:
                f.write(f"{book}\n")
        print(f"Books saved successfully to '{filename}'.")
    except OSError as e:
        print(f"Error saving books to file: {e}")


# TODO: Implement a function to read books from a file
def read_books_from_file(filename):
    """Read book titles from a text file and return them as a list."""
    try:
        with open(filename, "r") as f:
            # strip() removes the newline character from each line
            return [line.strip() for line in f if line.strip()]
    except FileNotFoundError:
        print(f"Error: The file '{filename}' was not found.")
        return []
    except OSError as e:
        print(f"Error reading file: {e}")
        return []

# Test file I/O
save_books_to_file("books.txt")
print("Books read from file:", read_books_from_file("books.txt"))
```

```python
# 7. Pickle for serialization
def save_books_pickle(filename):
    """Save the books list to a binary pickle file."""
    try:
        with open(filename, "wb") as f:
            pickle.dump(books, f)
        print(f"Books saved successfully to '{filename}'.")
    except OSError as e:
        print(f"Error saving pickle file: {e}")


# TODO: Implement a function to load books using pickle
def load_books_pickle(filename):
    """Load and return a books list from a pickle file."""
    try:
        with open(filename, "rb") as f:
            return pickle.load(f)
    except FileNotFoundError:
        print(f"Error: The file '{filename}' was not found.")
        return []
    except pickle.UnpicklingError:
        print(f"Error: '{filename}' is not a valid pickle file.")
        return []
    except OSError as e:
        print(f"Error loading pickle file: {e}")
        return []

# Test pickle functions
save_books_pickle("books.pkl")
print("Books loaded from pickle:", load_books_pickle("books.pkl"))
```

```python
# 8. SQLite Database Interaction
def create_books_table():
    """Create the books table if it does not already exist."""
    conn = None
    try:
        conn = sqlite3.connect("library.db")
        cursor = conn.cursor()
        cursor.execute("""
            CREATE TABLE IF NOT EXISTS books
            (id INTEGER PRIMARY KEY, title TEXT, quantity INTEGER)
        """)
        conn.commit()
        print("Books table created or already exists.")
    except sqlite3.Error as e:
        print(f"Database error while creating table: {e}")
    finally:
        if conn:
            conn.close()


# TODO: Implement a function to add a book to the SQLite database
def add_book_to_db(title, quantity):
    """Add a book title and quantity to the SQLite database."""
    # Validate the quantity before inserting it
    if not isinstance(quantity, int) or quantity < 0:
        print("Error: Quantity must be a non-negative integer.")
        return

    conn = None
    try:
        conn = sqlite3.connect("library.db")
        cursor = conn.cursor()

        # Use a parameterized query to prevent SQL injection
        cursor.execute(
            "INSERT INTO books (title, quantity) VALUES (?, ?)",
            (title, quantity)
        )
        conn.commit()
        print(f"Book '{title}' added to the database with quantity {quantity}.")
    except sqlite3.Error as e:
        print(f"Database error while adding book: {e}")
    finally:
        if conn:
            conn.close()

# Test database functions
create_books_table()
add_book_to_db("Python Programming", 5)
add_book_to_db("Advanced Python", 3)
add_book_to_db("Data Science with Python", 2)
```

```python
# Optional: Display all books stored in the SQLite database
def display_books_from_db():
    """Read and display all books stored in the database."""
    conn = None
    try:
        conn = sqlite3.connect("library.db")
        cursor = conn.cursor()
        cursor.execute("SELECT id, title, quantity FROM books")
        results = cursor.fetchall()

        print("\nBooks in the database:")
        if results:
            for book_id, title, quantity in results:
                print(f"ID: {book_id}, Title: {title}, Quantity: {quantity}")
        else:
            print("No books found in the database.")
    except sqlite3.Error as e:
        print(f"Database error while reading books: {e}")
    finally:
        if conn:
            conn.close()

display_books_from_db()
```

```python
# 9. Web API Interaction (GitHub API)
def get_python_repos():
    url = "https://api.github.com/search/repositories"
    params = {"q": "language:python", "sort": "stars", "order": "desc"}

    response = requests.get(url, params=params)
    data = response.json()

    return [repo["name"] for repo in data["items"][:5]]

# Test API function
print("Top 5 Python repositories on GitHub:", get_python_repos())

print("\nLibrary Management System operations completed.")
```
