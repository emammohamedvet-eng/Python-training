```python
# Task 1: Create a DataFrame with Name, Age, and City columns

import pandas as pd
import numpy as np

data = {
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Age": [25, 30, 35, 28],
    "City": ["New York", "San Francisco", "Los Angeles", "Chicago"]
}

# Create the DataFrame
df = pd.DataFrame(data)

# Check column names and data types
print("DataFrame:")
print(df)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)
```

```python
# Task 2: Drop the City column and verify the remaining columns

# Copy the original DataFrame so Task 1 remains unchanged
df_no_city = df.drop(columns=["City"])

# Verify that only Name and Age remain
print("DataFrame after dropping 'City':")
print(df_no_city)

print("\nRemaining columns:")
print(df_no_city.columns.tolist())

# Error handling: raise an error if the expected columns are missing
expected_columns = ["Name", "Age"]
if df_no_city.columns.tolist() != expected_columns:
    raise ValueError("The City column was not dropped correctly.")
else:
    print("\nThe DataFrame now contains only 'Name' and 'Age'.")
```

```python
# Task 3: Create a DataFrame with null values and handle them in different ways

df_null = pd.DataFrame({
    "Name": ["Alice", "Bob", None, "David"],
    "Age": [25, np.nan, 35, 28],
    "City": ["New York", None, "Los Angeles", "Chicago"]
})

print("Original DataFrame with null values:")
print(df_null)

print("\nNumber of null values in each column:")
print(df_null.isnull().sum())

# Method 1: Fill numeric null values with 0
df_filled_zero = df_null.copy()
df_filled_zero["Age"] = df_filled_zero["Age"].fillna(0)

# Method 2: Fill text null values with a meaningful label
df_filled_zero["Name"] = df_filled_zero["Name"].fillna("Unknown")
df_filled_zero["City"] = df_filled_zero["City"].fillna("Unknown")

print("\nDataFrame after filling null values:")
print(df_filled_zero)

# Method 3: Drop rows that contain any null values
df_dropped = df_null.dropna()

print("\nDataFrame after dropping rows with null values:")
print(df_dropped)

# Method 4: Fill numeric null values with the column mean
df_mean_filled = df_null.copy()
df_mean_filled["Age"] = df_mean_filled["Age"].fillna(df_mean_filled["Age"].mean())

print("\nDataFrame after filling Age null values with the mean:")
print(df_mean_filled)
```

```python
# Task 4: Group by Category and describe the Value column

df_group = pd.DataFrame({
    "Category": ["A", "B", "A", "B", "A", "C"],
    "Value": [10, 20, 15, 25, 30, 35]
})

# Group by Category and calculate descriptive statistics for Value
group_description = df_group.groupby("Category")["Value"].describe()

print("Descriptive statistics of Value by Category:")
print(group_description)

# Interpretation
print("\nInterpretation:")
print("- Category A has 3 values: 10, 15, and 30. Its mean is 18.33.")
print("- Category B has 2 values: 20 and 25. Its mean is 22.5.")
print("- Category C has 1 value: 35. Its mean is 35.")
print("- The standard deviation is 0 for Category C because it has only one value.")
```

```python
# Task 5: Concatenate df1 and df2 vertically, then combine with df3 horizontally

df1 = pd.DataFrame({"A": [1, 2], "B": [3, 4]})
df2 = pd.DataFrame({"A": [5, 6], "B": [7, 8]})
df3 = pd.DataFrame({"C": [9, 10], "D": [11, 12]})

# Vertically concatenate df1 and df2
df_concat = pd.concat([df1, df2], ignore_index=True)

print("DataFrame after vertical concatenation:")
print(df_concat)

# Horizontally combine the concatenated DataFrame with df3
# axis=1 keeps all rows from df_concat; missing values appear as NaN
df_horizontal = pd.concat([df_concat, df3], axis=1)

print("\nFinal DataFrame after horizontal concatenation:")
print(df_horizontal)

# Check the final shape: 4 rows and 4 columns
print("\nFinal DataFrame shape:", df_horizontal.shape)

# Error handling: verify that all original columns are present
expected_columns = ["A", "B", "C", "D"]

if df_horizontal.shape == (4, 4) and df_horizontal.columns.tolist() == expected_columns:
    print("\nThe final DataFrame has the correct shape and contains all columns.")
else:
    raise ValueError("The concatenated DataFrame has an unexpected shape or columns.")
```

```python
# Task 6: Demonstrate the difference between tuples and sets

# Create a tuple with at least 3 fruits
fruits = ("apple", "banana", "orange")
print("Original tuple:", fruits)

# Try to add an element to the tuple
try:
    fruits.add("mango")
except AttributeError as error:
    print("\nCannot add to a tuple:")
    print(error)

# Create a set with at least 5 numbers
numbers = {1, 2, 3, 4, 5}
print("\nOriginal set:", numbers)

# Add an element to the set
numbers.add(6)
print("Set after adding 6:", numbers)

# Explanation
print("\nExplanation:")
print("A tuple is immutable, so its elements cannot be added, removed, or changed after creation.")
print("A set is mutable, so new unique elements can be added using the add() method.")
```

```python
# Task 7: Create, update, and add to a dictionary

# Create a dictionary of student names and scores
students = {
    "Alice": 85,
    "Bob": 90,
    "Charlie": 78
}

print("Original dictionary:")
print(students)

# Update the score of an existing student
students["Alice"] = 92

# Add a new student
students["David"] = 88

print("\nDictionary after updating Alice and adding David:")
print(students)

# Error handling: safely retrieve a student's score
student_name = "Bob"

if student_name in students:
    print(f"\n{student_name}'s score is {students[student_name]}.")
else:
    print(f"\n{student_name} is not in the dictionary.")
```

```python
# Task 8: Calculate the square of a number using a function and a lambda function

# Regular function
def square(number):
    """Return the square of a number."""
    return number ** 2

# Lambda function
square_lambda = lambda number: number ** 2

# Test both functions with two different inputs
test_values = [4, 7]

for value in test_values:
    print(f"Regular function: square({value}) = {square(value)}")
    print(f"Lambda function: square_lambda({value}) = {square_lambda(value)}")
    print()
```

```python
# Task 9: Create an iterator class and a generator function for the first 5 even numbers

# Iterator class
class EvenNumbers:
    """Iterator that generates the first 5 even numbers."""

    def __init__(self):
        self.current = 2
        self.count = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= 5:
            raise StopIteration
        even_number = self.current
        self.current += 2
        self.count += 1
        return even_number


# Generator function
def even_numbers_generator():
    """Generator that yields the first 5 even numbers."""
    for number in range(2, 11, 2):
        yield number


# Use the iterator
print("Iterator output:")
for number in EvenNumbers():
    print(number)

# Use the generator
print("\nGenerator output:")
for number in even_numbers_generator():
    print(number)

# Convert the generator output to a list for display
print("\nGenerator as a list:")
print(list(even_numbers_generator()))
```

```python
# Task 10: Use map(), reduce(), and filter() on a list of numbers

from functools import reduce

numbers = [1, 2, 3, 4, 5]

# Use map() to square every number
squared_numbers = list(map(lambda x: x ** 2, numbers))

# Use reduce() to calculate the product of all numbers
product = reduce(lambda x, y: x * y, numbers)

# Use filter() to select only even numbers
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

print("Original numbers:", numbers)
print("Squared numbers using map():", squared_numbers)
print("Product of all numbers using reduce():", product)
print("Even numbers using filter():", even_numbers)
```

```python
# Task 11: Create a Rectangle class with area and perimeter methods

class Rectangle:
    """A class representing a rectangle."""

    def __init__(self, length, width):
        # Validate that dimensions are positive numbers
        if length <= 0 or width <= 0:
            raise ValueError("Length and width must be positive numbers.")

        self.length = length
        self.width = width

    def calculate_area(self):
        """Return the area of the rectangle."""
        return self.length * self.width

    def calculate_perimeter(self):
        """Return the perimeter of the rectangle."""
        return 2 * (self.length + self.width)


# Create two Rectangle objects
rectangle1 = Rectangle(5, 3)
rectangle2 = Rectangle(10, 4)

# Demonstrate the methods
print("Rectangle 1:")
print("Length:", rectangle1.length)
print("Width:", rectangle1.width)
print("Area:", rectangle1.calculate_area())
print("Perimeter:", rectangle1.calculate_perimeter())

print("\nRectangle 2:")
print("Length:", rectangle2.length)
print("Width:", rectangle2.width)
print("Area:", rectangle2.calculate_area())
print("Perimeter:", rectangle2.calculate_perimeter())
```

```python
# Task 12: Analyze employee salary data using Pandas

df_employees = pd.DataFrame({
    "Name": ["John", "Jane", "Bob", "Alice", "Charlie"],
    "Department": ["IT", "HR", "IT", "Finance", "HR"],
    "Salary": [55000, 65000, 70000, 60000, 58000]
})

# 1. Calculate average salary by department
average_salary = df_employees.groupby("Department")["Salary"].mean()

print("Average salary by department:")
print(average_salary)

# 2. Extract names of employees with salary greater than 60000
high_salary_employees = df_employees.loc[
    df_employees["Salary"] > 60000, "Name"
]

print("\nEmployees with salary greater than 60000:")
print(high_salary_employees.tolist())

# 3. Add a Bonus column equal to 10% of salary
df_employees["Bonus"] = df_employees["Salary"] * 0.10

print("\nEmployee DataFrame with Bonus column:")
print(df_employees)

# Verify that the Bonus column was added correctly
if "Bonus" in df_employees.columns:
    print("\nThe Bonus column was added successfully.")
else:
    raise ValueError("The Bonus column was not created.")
```
